```python
l1=[1,2,3,4,5]
l2=[5,6,7,8,9]
prod=[]
for i in range(len(l1)):
    print(l1[i]*l2[i])
    prod.append(l1[i]*l2[i])
print(prod)


```

```python
import numpy as np
#craete numpy array

a=np.array([1,2,3,4,5])
print("Array:",a)
print("Type:",type(a))
print("Data Type:",a.dtype)
print("Shape:",a.shape)
print("Size:",a.size)
print("Dimension:",a.ndim)   
print()
```

```python
arr1=np.array([1,2,3,4,5])
arr2=np.arange(1,6)
arr3=np.zeros(5)
arr4=np.ones(5)
arr5=np.linspace(0,10,5)
arr6=np.empty(6)

print("From List",arr1)
print("Range:",arr2)
print("Zeros:",arr3)
print("Ones:",arr4)
print("Linspace:",arr5)
print("Empty:",arr6)
```
