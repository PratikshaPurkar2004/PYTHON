```python
import numpy as np
```

```python
def dataset(rows=10,cols=5):
    return np.random.randint(1,101,size=(rows,cols))
```

```python
def analyze_dataset(data):
    print(data)
    All_mean=np.mean(data)
    print("overAll_mean=",All_mean)
    print("Shape ::",data.shape)
    print("Dimensions ::",data.ndim)
    print("Variable_mean ::",np.mean(data,axis=0))
    print("Experiment_mean ::",np.mean(data,axis=1))
    print("maximum",np.max(data))
    print("minimum",np.min(data))
    print("above_mean",data[data>All_mean])
    reshaped = data.reshape(5, 10)
    print("reshaped data::\n",reshaped)
    flattened = reshaped.flatten()
    print("flattend::\n",flattened)
    
```

```python
data=dataset()
analyze_dataset(data)
```
