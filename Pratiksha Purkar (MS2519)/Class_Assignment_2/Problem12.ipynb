```python
import numpy as np
```

```python
image = np.array([
[0, 0, 0, 255, 255],
[0, 0, 128, 255, 255],
[0, 128, 128, 128, 255],
[0, 0, 128, 0, 0],
[255, 0, 0, 0, 255]
])

print(image)
```

```python
minimum=np.min(image)
print(minimum)
```

```python
maximum=np.max(image)
print(maximum)
```

```python
mean=np.mean(image)
print(mean)
```

```python
greater_than_100=image[image>100]
print(greater_than_100)
```

```python
cnt_greater_than_100=np.count_nonzero(image>100)
print(cnt_greater_than_100)
```

```python
pixels_255=image[image==255]
print(pixels_255)
```
