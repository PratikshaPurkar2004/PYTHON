```python
import numpy as np
```

```python
measurment=np.array([10, 12, 15, 200, 13, 11, 14, 250, 12, 15])
print(measurment)
```

```python
invalid=measurment[measurment>100]
print(invalid)
```

```python
valid=measurment[measurment<=100]
print(valid)
```

```python
mean_valid=np.mean(valid)
print(mean_valid)
```

```python
minimum_valid=np.min(valid)
print(minimum_valid)
```

```python
maximum_valid=np.max(valid)
print(maximum_valid)
```
