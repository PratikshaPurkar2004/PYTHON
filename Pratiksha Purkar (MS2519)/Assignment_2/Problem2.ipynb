```python
str=input("Enter a sentences::")
print(str)
words=len(str.split()) #no of words
sentences=len(str) #no of character
vowel=0
consonent=0
for ch in str:
    if ch in "aeiouAEIOU":
        vowel+=1
    else:
        consonent+=1
print("Number of words::",words)
print("Number of sentences::",sentences)
print("Number of vowels::",vowel)
print("Number of consonents::",consonent)
```
