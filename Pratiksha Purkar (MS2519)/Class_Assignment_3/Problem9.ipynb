```python
class BankAccount:
    def __init__(self,accNum,owner,bal):
        self.accNum = accNum
        self.owner = owner
        self.bal = bal

    def deposit(self,amount):
        self.bal += amount

    def withdraw(self,amount):
        if amount <= self.bal:
            self.bal -= amount
            return True
        else:
            return False

    def __str__(self):
        return "Account Number:"+str(self.accNum)+"\nOwner:"+self.owner+"\nBalance:"+str(self.bal)

acc=BankAccount(101,"John Doe",1000)
print("Initial balance:",acc.bal)
acc.deposit(2000)
print("Balance after deposit:",acc.bal)

result1= acc.withdraw(1000)
print("Successful withdrawal of 1000:",result1)
print("Balance after withdrawal:",acc.bal)

print(acc)
```
