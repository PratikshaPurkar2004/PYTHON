```python
class BankAccount:

    def __init__(self, account_num, owner, balance):
        self.account_num = account_num
        self.owner = owner
        self.balance = balance

    def __str__(self):
        return "Account Number: " + str(self.account_num) + \
               ", Owner: " + self.owner + \
               ", Balance: " + str(self.balance)

account = BankAccount(101, "Pratiksha", 5000)

print(account)
```
