```python
class Car:
    def move(self):
        print("car is driving on the road")

class Boat:
    def move(self):
        print("boat is sailing on the water")

class Airplane:
    def move(self):
        print("airplane is flying in the sky")


car1=Car()
boat1=Boat()
airplane1=Airplane()

vehicles=[car1,boat1,airplane1]

for vehicle in vehicles:
    vehicle.move()
```
