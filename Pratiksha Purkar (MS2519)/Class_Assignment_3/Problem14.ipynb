```python
class Vehicle:
    def __init__(self,brand):
        self.brand =brand

class Car(Vehicle):
    def __init__(self,brand,doores):
        super().__init__(brand)
        self.doores=doores

car1=Car("Toyota",4)
car2=Car("Honda",2)

print("Car 1\n")
print(car1.brand,car1.doores)
print("Car 2\n")
print(car2.brand,car2.doores)
```
