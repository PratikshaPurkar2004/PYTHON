```python
class BankAccount:
    def __init__(self,acc_num,owner,balance):
        self.acc_num=acc_num
        self.owner=owner
        self.balance=balance
    def deposit(self,amt):
        self.balance+=amt
    def withdraw(self,amt):
        if(amt<self.balance):
             self.balance-=amt
        else:
            print("Insufficient balance")
```

```python
acc=BankAccount("1234567","John Doe",1000.0)
acc.deposit(500)
print(acc.balance)
acc.withdraw(2000)
print(acc.balance)
```
