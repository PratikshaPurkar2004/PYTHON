```python
import pandas as pd
```

```python
df=pd.read_csv("online food delivery dataset.csv")
```

```python
feedback_count=df['Feedback'].value_counts()
print(feedback_count)
```

```python
positive_feedback_count=feedback_count.get('Positive', 0)
print(positive_feedback_count)
total=len(df)
percentage_positive=(positive_feedback_count/total)*100
print(percentage_positive)
```
