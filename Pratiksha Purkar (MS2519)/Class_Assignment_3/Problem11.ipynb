```python
class Student:
    def __init__(self,name ,rollNo,marks):
        self.name=name
        self.rollNo=rollNo
        self.marks=marks

    def average_marks(self):
        return sum(self.marks)/len(self.marks)
```

```python
s1=Student('kavya',1,[98,45,76])
s2=Student('Nirmala',3,[85,65,89])
s3=Student('Sanskruti',9,[76,66,54])
```

```python
print("student 1")
print(s1.name)
print(s1.rollNo)
print(s1.average_marks())
```

```python
print("student 2")
print(s2.name)
print(s2.rollNo)
print(s2.average_marks())
```
