```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import BarycentricInterpolator

# --------------------------------
# 1. Load dataset
# --------------------------------
data = pd.read_csv("housing.csv")

x_all = data["median_income"].values
y_all = data["median_house_value"].values / 100000

# Random seed
np.random.seed(42)


# --------------------------------
# 2. Function for Lagrange graph
# --------------------------------
def plot_lagrange(n):

    # Select random n datapoints
    indices = np.random.choice(len(x_all), n, replace=False)

    x = x_all[indices]
    y = y_all[indices]

    # Sort according to x
    order = np.argsort(x)
    x = x[order]
    y = y[order]

    # Create Lagrange polynomial
    polynomial = BarycentricInterpolator(x, y)

    # Generate x values for graph
    x_plot = np.linspace(x.min(), x.max(), 500)

    # Calculate polynomial values
    y_plot = polynomial(x_plot)

    # Plot
    plt.figure(figsize=(8, 5))

    plt.scatter(
        x,
        y,
        label=f"{n} selected datapoints"
    )

    plt.plot(
        x_plot,
        y_plot,
        label=f"Degree {n-1} Lagrange polynomial"
    )

    plt.xlabel("Median Income")
    plt.ylabel("Median House Value")
    plt.title(
        f"Lagrange Interpolation using {n} Datapoints"
    )

    plt.legend()
    plt.grid(True)

    plt.show()


# --------------------------------
# 3. Plot for 50 datapoints
# --------------------------------

plot_lagrange(50)


# --------------------------------
# 4. Plot for 100 datapoints
# --------------------------------

plot_lagrange(100)


# --------------------------------
# 5. Plot for 500 datapoints
# --------------------------------

plot_lagrange(500)
```
