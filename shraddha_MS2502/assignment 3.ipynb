```python
#2
import pandas as pd
df = pd.read_csv("data.csv", sep="\t")
df.columns = df.columns.str.strip()

print("Average age of the customers:")
print(df["Age"].mean())

print("\nMinimum age of the customers:")
print(df["Age"].min())

print("\nMaximum age of the customers:")
print(df["Age"].max())

print("\nNumber of male customers:")
print((df["Gender"].str.strip() == "Male").sum())

print("\nNumber of female customers:")
print((df["Gender"].str.strip() == "Female").sum())
```

```python
##4
import pandas as pd
df = pd.read_csv("data.csv", sep="\t")
average_age = df.groupby("Occupation")["Age"].mean()
print("Average age for each occupation:")
print(average_age)

highest_average_age = average_age.idxmax()
print("\nOccupation with the highest average customer age:")
print(highest_average_age)

print("\nHighest average age:")
print(average_age.max())
```

```python
##8
class BankAccount:
    def __init__(self, account_number, owner, balance):
        self.account_number = account_number
        self.owner = owner
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds")
        else:
            self.balance -= amount
account=BankAccount(101, "Shraddha", 1000)

print("initial balance:", account.balance)
account.deposit(500)

print("balance after deposit:", account.balance)
account.withdraw(200)
print("balance after withdrawal:", account.balance)
```

```python
##9
class BankAccount:
    def __init__(self, account_number, owner, balance):
        self.account_number = account_number
        self.owner = owner
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount < self.balance:
            self.balance -= amount
            return True
        else:
            return False

    account3=BankAccount(103,"Rohit", 1500)
    result = account3.withdraw(1000)
    print("Withdrawal successful:", result)
    print("Balance:", account3.balance)

    result = account3.withdraw(5000)
    print("Withdrawal successful:", result)
    print("Balance:", account3.balance)
```

```python
##11
class student:
    def __init__(self, name, rollno,marks):
        self.name=name
        self.rollno=rollno
        self.marks=marks

    def average_marks(self):
        return sum(self.marks) / len(self.marks)

student1=student("Shraddha", 101, [85,70,49,90])
student2=student("Priti", 102, [75,80,60,90])
student3=student("Pratiksha", 103, [90,85,70,95])

print("student1:",student1.name)
print("average:",student1.average_marks())

print("\nstudent2:",student2.name)
print("average:",student2.average_marks())

print("\nstudent3:",student3.name)
print("average:",student3.average_marks())
```

```python
##14
class vehicle:
  def __init__(self,brand):
    self.brand = brand

class car(vehicle):
  def __init__(self, brand, doors):
    super().__init__(brand)
    self.doors = doors

car1 = car("Scorpio", 4)
print("Car brand:", car1.brand)
print("Car doors:", car1.doors)

car2 = car("Fortuner", 5)
print("Car brand:", car2.brand)
print("Car doors:", car2.doors)
    
```

```python
##15
class car:
    def move(self):
        print("car is moving")

class boat:
    def move(self):
        print("boat is sailing")

class airplane:
    def move(self):
        print("aeroplane is flying")

car=car()
boat=boat()
airplane=airplane()

vehicles=[car,boat,airplane]
for vehicle in vehicles:
    vehicle.move()
```
