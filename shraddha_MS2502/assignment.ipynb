```python
#list methods
#adds one item at the end
list=[10,20,30,38]
list.append(10)
print(list)
```

```python
#adds the item at specific position
fruit=["apple","banana","cherry"]
fruit.insert("orange")
print(fruit)
```

```python
#removes the item from the list
Name=["shraddha","pratiksha","prapti","saily","aarya","sanika"]
Name.remove("aarya")
print(Name)
```

```python
#remove an tem by index
city=["pune","mumbai","nagpur","nashik"]
city.pop(2)
print(city)
```

```python
#sorts the list in ascending order
numbers=[7,4,6,3,2,6,9,1]
numbers.sort()
print(numbers)
```

```python
#Tuple
#counts how many times an itemappears
numbers=(1,5,3,4,5,6,5,8,9,10)
print(numbers.count(5))
```

```python
#returns the position of an item
colors=("red","green","blue","yellow")
print(colors.index("blue"))
```

```python
#returs total number of elements or items
country=("india","france","america","england")
print(len(country))
```

```python
#return larget value
n=(10,400,333,78999,654)
print(max(n))
```

```python
#returns the smallest value
m=(10,2,4,1,8,444,16)
print(min(m))
```

```python
#set
#adds the items
num={1,2,3,4,5,6,7,8,9}
num.add(16)
print(num)
```

```python
#remove an item
numbers={10,20,50,60,80,70,100}
numbers.remove(100)
print(numbers)
```

```python
#remove an itemif exists
num={10,60,48,49,20,16,25}
num.discard(20)
print(num)
```

```python
#combines two sets
set1={10,20,30,40}
set2={50,60,70,80}
print(set1.union(set2))
```

```python
#return common items
a={1,3,6,8,9}
b={2,4,6,8,10}
print(a.intersection(b))
```

```python
#dictonary
#gets the value of key
student={"Name":"shraddha","Name":"durgesh","Name":"ridhima","Name":"nakul"}
print(student.get("Name"))
```

```python
#return all keys
student={"Name":"shraddha","Age":20,"City":"pune"}
print(student.keys())
```

```python
#returns all values
student={"Name":"shraddha","Age":20,"City":"pune"}
print(student.values())
```

```python
#returns key-values  pairs
student={"Name":"shraddha","Age":20,"City":"pune"}
print(student.items())
```
