```python
#2
import numpy as np
def temperature_analysis(temperatures):
    average = np.mean(temperatures)
    maximum = np.max(temperatures)
    minimum = np.min(temperatures)
    highest_day = np.argmax(temperatures) + 1
    above_average = temperatures[temperatures > average]
    print("Average temperature:", average)
    print("Maximum temperature:", maximum)
    print("Minimum temperature:", minimum)
    print("Day with highest temperature:", highest_day)
    print("Temperatures above average:", above_average)
temperatures = np.array([
    32.5, 31.8, 30.2, 33.1, 34.5, 29.8, 31.0
])
temperature_analysis(temperatures)
```

```python
#6
import numpy as np
marks = np.array([
    [78, 82, 91],
    [65, 74, 80],
    [92, 88, 95],
    [56, 69, 72],
    [84, 90, 87]
])
print("Student marks:",marks)
student_average = np.mean(marks, axis=1)
subject_average = np.mean(marks, axis=0)
highest_mark = np.max(marks)
lowest_mark = np.min(marks)
best_student = np.argmax(student_average)
subjects = [
    "Mathematics",
    "Physics",
    "Programming"
]
best_subject = np.argmax(subject_average)
print("\nAverage of each student:",student_average)
print("\nAverage of each subject:",subject_average)
print("\nHighest mark:",highest_mark)
print("\nLowest mark:",lowest_mark)
print("\nStudent with highest average:", best_student + 1)
print("\nSubject with highest average:",subjects[best_subject])
```

```python
#8
import numpy as np
def transform_data(data):
    square = data ** 2
    square_root = np.sqrt(data)
    exponential = np.exp(data)
    print("Original data:",data)
    print("\nSquare:",square)
    print("\nSquare root:",square_root)
    print("\nExponential:",exponential)
    print("\nMean of square:",np.mean(square))
    print("\nMean of square root:",np.mean(square_root))
    print("\nMean of exponential:",np.mean(exponential))
data = np.array([2, 4, 6, 8, 10, 12])
transform_data(data)
```

```python
#12
import numpy as np
image = np.array([
    [0, 0, 0, 255, 255],
    [0, 0, 128, 255, 255],
    [0, 128, 128, 128, 255],
    [0, 0, 128, 0, 0],
    [255, 0, 0, 0, 255]
])
minimum = np.min(image)
maximum = np.max(image)
mean = np.mean(image)
above_100 = image[image > 100]
count = len(above_100)
pixels_255 = image[image == 255]
print("Minimum pixel intensity:",minimum)
print("\nMaximum pixel intensity:",maximum)
print("\nMean pixel intensity:",mean)
print("\nNumber of pixels greater than 100:",count)
print("\nPixels equal to 255:",pixels_255)
print("\nNew array containing pixels greater than 100:",above_100)
```

```python
#13
import numpy as np
experiment_A = np.array([
    12, 15, 14, 18, 20, 17, 16
])
experiment_B = np.array([
    10, 13, 15, 16, 19, 21, 18
])
mean_A = np.mean(experiment_A)
std_A = np.std(experiment_A)
max_A = np.max(experiment_A)
min_A = np.min(experiment_A)

mean_B = np.mean(experiment_B)
std_B = np.std(experiment_B)
max_B = np.max(experiment_B)
min_B = np.min(experiment_B)
print("Experiment A")
print("Mean:", mean_A)
print("Standard deviation:", std_A)
print("Maximum:", max_A)
print("Minimum:", min_A)

print("\nExperiment B")
print("Mean:", mean_B)
print("Standard deviation:", std_B)
print("Maximum:", max_B)
print("Minimum:", min_B)

difference = experiment_A - experiment_B
print("\nDifference between experiments:",difference)

if std_A > std_B:
    print("\nExperiment A has greater variation.")
elif std_B > std_A:
    print("\nExperiment B has greater variation.")
else:
    print("\nBoth have the same variation.")
```

```python
#14
import numpy as np
def create_data():
    data = np.random.randint(1, 101,(10, 5))
    return data

def analyze_data(data):
    print("Dataset:",data)
    print("\nShape:",data.shape)
    print("\nNumber of dimensions:",data.ndim)
    variable_mean = np.mean(data,axis=0)
    print("\nMean of each variable:",variable_mean)
    experiment_mean = np.mean(data,axis=1)
    print("\nMean of each experiment:",experiment_mean)
    print("\nMaximum value:",np.max(data))
    print("\nMinimum value:",np.min(data))
    overall_mean = np.mean(data)
    print("\nOverall mean:",overall_mean)
    above_mean = data[data > overall_mean]
    print("\nValues above overall mean:",above_mean)
    reshaped = data.reshape(5, 10)
    print("\nReshaped data:",reshaped)
    print("New shape:",reshaped.shape)
    flattened = reshaped.flatten()
    print("\nFlattened data:",flattened)
data = create_data()
analyze_data(data)
```
