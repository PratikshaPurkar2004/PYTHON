```python
import numpy as np
arr=np.arange(12)
print(f"original array:\n{arr}")
matrix=arr.reshape(3,4)
f=matrix.flatten()
r=matrix.ravel()
#print(f"flattened array:\n{matrix.flatten()}")
#print(f"Raveled array:\n{matrix.ravel()}")
f[0]=100
print(f"flattened array:\n{f}")
print(f"flattened array:\n{matrix}")
f[0]=200
print(f"Raveled array:\n{r}")
print(f"Raveled array:\n{matrix}")
```
