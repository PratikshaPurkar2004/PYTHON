```python
list1=['Pratiksha','Shraddha','Sayli','Aarya']
print(list1)
```

```python
#List Method

#1)Append:Add Element at the end

list1.append('Harshda')
print(list1)
```

```python
#2) insert: Add Element at the specific index

list1.insert(2,'Rutuja')
print(list1)
```

```python
#3) extend: Add multiple elements at the end

list1.extend(['Isha', 'Kavya'])
print(list1)
```

```python
#4) remove: Remove an element from the list

list1.remove('Sayli')
print(list1)
```

```python
#5) pop: Remove an element at a specific index

list1.pop(3)
print(list1)
```

```python
#6)sort: Sort the list in ascending order

list.sort(list1)
print(list1)

#descending order
list.sort(list1, reverse=True)
print(list1)
```

```python
#7)reverse: reverse list

print(list1.reverse())
#
```

```python
# Tuple Methods

tuple1=(1,2,3,4,5,2)

#1) count: Count the number of occurrences of an element in a tuple
print(tuple1.count(2))
```

```python
#2)index: Return the index of the first occurrence of an element in a tuple
print(tuple1.index(3))
```

```python
#3)len: Return the length of a tuple
print(len(tuple1))
```

```python
#4)max: Return the maximum value in a tuple
print(max(tuple1))
```

```python
#5)min: Return the minimum value in a tuple
print(min(tuple1))
```

```python
#Set Methods

set1={1,2,3,4,5,6}
print(set1)
```

```python
#1)add: Add an element to a set
set1.add(7)
print(set1)
```

```python
#2)remove: Remove an element from a set
set1.remove(3)
print(set1)
```

```python
#3)update: Add multiple elements to a set
set1.update([8,9,10])
print(set1)
```

```python
#4)discard: Remove an element from a set if it exists
set1.discard(5)
print(set1)
```

```python
#5)clear: Remove all elements from a set
set1.clear()
print(set1)
```

```python
#6)union: Return a new set with all elements from both sets

set1={1,2,3,4}
set2={4,5,6}
set3=set1.union(set2)
print(set3)
```

```python
#7)intersection: Return a new set with elements common to both sets
set3=set1.intersection(set2)
print(set3)
```

```python
#8)difference: Return a new set with elements in the first set but not in the second set
set3=set1.difference(set2)
print(set3)
```

```python
#Dictionary Methods

dict1={'name':'Pratiksha','age':21,'city':'Pune'}
print(dict1)
```

```python
#1)keys: Return a view object that displays a list of all the keys in the dictionary
print(dict1.keys())
```

```python
#2)values: Return a view object that displays a list of all the values in the dictionary
print(dict1.values())
```

```python
#3)items: Return a view object that displays a list of all the key-value pairs in the dictionary
print(dict1.items())
```

```python
#4)get: Return the value of a specified key in the dictionary
print(dict1.get('name'))
```

```python
#5)update: Update the value of a specified key in the dictionary
dict1.update({'age':22})
print(dict1)
```
