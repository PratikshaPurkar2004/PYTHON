```python
list1=['Pratiksha','Shraddha','Sayli','Aarya']
print(list1)
```

```python
#List Method

#1)Append:Add Element at the end

list1.append('Harshda')
print(list1)
```

```python
#2) insert: Add Element at the specific index

list1.insert(2,'Rutuja')
print(list1)
```

```python
#3) extend: Add multiple elements at the end

list1.extend(['Isha', 'Kavya'])
print(list1)
```

```python
#4) remove: Remove an element from the list

list1.remove('Sayli')
print(list1)
```

```python
#5) pop: Remove an element at a specific index

list1.pop(3)
print(list1)
```

```python
#6)sort: Sort the list in ascending order

list.sort(list1)
print(list1)

#descending order
list.sort(list1, reverse=True)
print(list1)
```

```python
#7)reverse: reverse list

print(list1.reverse())
#
```
