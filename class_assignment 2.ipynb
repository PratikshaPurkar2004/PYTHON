```python
marks=[]
for i in range(5):
    #mark=int(input("Enter marks:"+str(i+1)+":"))
    mark=int(input("Enter marks:"))
    marks.append(mark)
    print(marks)
    average=sum(marks)/5
    print("marks:",marks)
    print("Average:", average)
    if average>=90:
        print("Grade: A")
    elif average>=80:
        print("Grade: B")   
    elif average>=70:
        print("Grade: C")
    elif average>=60:
        print("Grade: D")
    elif average>=50:
        print("Grade: E")
    else:
        print("Grade: F")
```

```python
sentence=input("Enter a sentence:")
words=len(sentence.split())
characters=len(sentence)
vowels=0
consonants=0
for char in sentence:
    if char.lower() in "aeiou":
        vowels+=1
    elif char.isalpha():
        consonants+=1
print("Number of words:", words)
print("Number of characters:", characters)
print("Number of vowels:", vowels)
print("Number of consonants:", consonants)
```

```python
numbers=[]
odd=[]
even=[]
num=int(input("how many numbers do u want enter:"))
for i in range(num):
    number=int(input("Enter a number:"))
    numbers.append(number)
    if number%2==0:
        even.append(number)
    else:
        odd.append(number)
print("odd number:",odd)
print("even number:",even)
```

```python
numbers=list(map(int,input("enter the numbers:").split()))
largest=numbers[0]
smallest=numbers[0]
total=0
for num in numbers:
    if num > largest:
        largest=num
    if num < smallest:
        smallest=num
    total=total+num
    average=total/len(numbers)
print("largest:",largest)
print("smallest:",smallest)
print("sum of elements:",total)
print("average:",average)

```

```python
marks=tuple(map(int,input("enter the marks:").split()))
highest=max(marks)
lowest=min(marks)
average=sum(marks)/len(marks)
count=0
for mark in marks:
    if mark>average:
        count=count+1
print("Highest mark:",highest)
print("Lowest mark:",lowest)
print("class average:",average)
print("student above average:",count)

```

```python
contacts = {}

while True:

    print("\n1.Add")
    print("2.Search")
    print("3.Update")
    print("4.Delete")
    print("5.Display")
    print("6.Exit")

    choice = int(input("Enter Choice: "))

    if choice == 1:

        name = input("Enter Name: ")
        number = input("Enter Number: ")

        contacts[name] = number

    elif choice == 2:

        name = input("Enter Name: ")

        if name in contacts:
            print("Number =", contacts[name])

        else:
            print("Contact Not Found")

    elif choice == 3:

        name = input("Enter Name: ")

        if name in contacts:

            number = input("Enter New Number: ")

            contacts[name] = number

            print("Updated")

        else:

            print("Contact Not Found")

    elif choice == 4:

        name = input("Enter Name: ")

        if name in contacts:

            del contacts[name]

            print("Deleted")

        else:

            print("Contact Not Found")

    elif choice == 5:

        print("Contacts")

        for name in contacts:
            print(name, ":", contacts[name])

    elif choice == 6:

        print("Thank You")

        break

    else:

        print("Invalid Choice")
```

```python
students = {}
while True:

    print("\n Student Record Management")
    print("1. Add Student")
    print("2. Search Student")
    print("3. Update Student")
    print("4. Delete Student")
    print("5. Display All Students")
    print("6. Exit")

    choice = int(input("Enter your choice: "))
    if choice == 1:
        roll = input("Enter Roll Number: ")
        name = input("Enter Name: ")
        course = input("Enter Course: ")
        marks = input("Enter Marks: ")
        students[roll] = {
            "name": name,
            "course": course,
            "marks": marks
        }
        print("Student added successfully.")
    elif choice == 2:
        roll = input("Enter Roll Number: ")
        if roll in students:
            print("Name:", students[roll]["name"])
            print("Course:", students[roll]["course"])
            print("Marks:", students[roll]["marks"])
        else:
            print("Student not found.")
    elif choice == 3:
         roll = input("Enter Roll Number: ")
         if roll in students:

            students[roll]["name"] = input("Enter New Name: ")
            students[roll]["course"] = input("Enter New Course: ")
            students[roll]["marks"] = input("Enter New Marks: ")

            print("Student updated successfully.")
         else:
            print("Student not found.")
    elif choice == 4:
        roll = input("Enter Roll Number: ")
        if roll in students:
            del students[roll]
            print("Student deleted successfully.")
        else:
            print("Student not found.")
    elif choice == 5:
        if len(students) == 0:
            print("No student records.")

        else:
            for roll in students:
                print("\nRoll Number:", roll)
                print("Name:", students[roll]["name"])
                print("Course:", students[roll]["course"])
                print("Marks:", students[roll]["marks"])
    elif choice == 6:
            print("Program ended.")
    break

else:
            print("Invalid choice.")
```
