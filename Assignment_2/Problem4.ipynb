```python

list1=list(map(int,input("Enter a numbers::").split()))
    
print(list1)

max=list1[0]
for i in list1:
    if max<i:
        max=i

min=list1[0]
for i in list1:
    if min>i:
        min=i

sum=0
for i in list1:
    sum+=i

avg=sum/len(list1)

print("Largest Element::",max)
print("Smallest Element::",min)
print("Sum of all numbers::",sum)
print("Average::",avg)
```
