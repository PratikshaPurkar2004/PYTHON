```python
import numpy as np


def transform_data(data):
    squares = np.square(data)
    square_roots = np.sqrt(data)
    exponentials = np.exp(data)

    means = {
        "square_mean": np.mean(squares),
        "sqrt_mean": np.mean(square_roots),
        "exponential_mean": np.mean(exponentials)
    }

    return squares, square_roots, exponentials, means


data = np.array([2, 4, 6, 8, 10, 12])

squares, roots, exponentials, means = transform_data(data)

print("Original Data:")
print(data)

print("\nSquares:")
print(squares)

print("\nSquare Roots:")
print(roots)

print("\nExponentials:")
print(exponentials)

print("\nMean of Squares:", means["square_mean"])
print("Mean of Square Roots:", means["sqrt_mean"])
print("Mean of Exponentials:", means["exponential_mean"])
```
