```python
import numpy as np

numbers = np.random.randint(1, 101, 30)
print(numbers)
```

```python
greater_50 = numbers[numbers > 50]
print(greater_50)
```

```python
between = numbers[(numbers >= 25) & (numbers <= 75)]
print(between)
```

```python
count = np.sum(numbers > 50) # Count numbers greater than 50
print(count)
```

```python
average = np.mean(greater_50)
print(average)
```
