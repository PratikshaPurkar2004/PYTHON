```python
import numpy as np
```

```python
data = np.arange(1, 25)
```

```python
array_4x6 = data.reshape(4, 6)
print(array_4x6)
print(array_4x6.shape)
```

```python
array_6x4 = data.reshape(6, 4)
print(array_6x4)
print(array_6x4.shape)
```

```python
array_2x3x4 = data.reshape(2, 3, 4)
print(array_2x3x4)
print(array_2x3x4.shape)
```
