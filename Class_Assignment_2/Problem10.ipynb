```python
import numpy as np
```

```python
experiment_1 = [10, 20, 30, 40, 50]
experiment_2 = [15, 25, 35, 45, 55]
```

```python
combine_1D=np.concatenate((experiment_1,experiment_2))
print(combine_1D)
print(combine_1D.shape)
```

```python
combine_row=np.vstack((experiment_1,experiment_2))
print(combine_row)
print(combine_row.shape)
```

```python
combine_col=np.column_stack((experiment_1,experiment_2))
print(combine_col)
print(combine_col.shape)
```
