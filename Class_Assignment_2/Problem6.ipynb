```python
import numpy as np
```

```python
students=np.array([[78,82,91],[65,74,80],[92,88,95],[56,69,72],[84,97,87]])
student_avg=np.mean(students,axis=1)
print(student_avg)
```

```python
sub_avg=np.mean(students,axis=0)
print(sub_avg)
```

```python
highest_mark = np.max(students)
print(highest_mark)
```

```python
lowest_mark = np.min(students)
print(lowest_mark)
```

```python
highest_student = np.argmax(student_avg) + 1
print(highest_student)
```

```python
subjects = ["Mathematics", "Physics", "Programming"]
highest_subject = students[np.argmax(sub_avg)]
print(highest_subject)
```
